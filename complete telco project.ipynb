# Feature Engineering

```python
# Feature 1 — Tenure Group
```

```python
df["Tenure_Group"]=pd.cut(
    df["Tenure Months"],
    bins=[-1, 6, 12, 24, 48, 72],
    labels=[
        "0-6 Months",
        "7-12 Months",
        "13-24 Months",
        "25-48 Months",
        "49-72 Months",
    ])
```

```python
# Customers with short tenure may have different churn behavior from long-term customers.
```

# Feature 2 — Average Monthly Spending

```python
df["Average_Monthly_Spending"]=(
    df["Total Charges"]/
    df["Tenure Months"].replace(0,np.nan))
```

```python
# Fill missing values
```

```python
df["Average_Monthly_Spending"]=(
    df["Average_Monthly_Spending"].replace([np.inf, -np.inf],np.nan))
```

```python
# It estimates how much a customer has spent per month
```

# Feature 3 — Customer Lifetime Value

```python
df["Customer_Lifetime_Value"]=df["CLTV"]
```

```python
# It represents the customer's estimated lifetime value.
```

# Feature 4 — Total Service Count

```python
service_columns = [
    "Phone Service",
    "Multiple Lines",
    "Online Security",
    "Online Backup",
    "Device Protection",
    "Tech Support",
    "Streaming TV",
    "Streaming Movies"
]

df["Total_Service_Count"] = (
    df[service_columns]
    .apply(lambda row: sum(
        1 for value in row
        if value in ["Yes", "Yes"]
    ), axis=1)
)
```

# Feature 5 — Service Usage Indicator

```python
df["Service_Usage_Indicator"]=(
    df["Total_Service_Count"]>2
).astype(int)
```

# Feature 6 — Monthly Charge Category

```python
df["Monthly_Charge_Category"]=pd.cut(
    df["Monthly Charges"],
    bins=[-np.inf, 40, 70, np.inf],
    labels=["Low", "Medium", "High"]
    )
```

```python
# def detect_outliers_iqr(data, column)
def detect_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3-Q1

    lower = Q1-1.5*IQR
    upper = Q3+1.5*IQR
    return data[
        (data[column]<lower) |
        (data[column]>upper)
    ]
```

# Separate X and y

```python
X=df.drop(columns=["Churn Label"])
y=df["Churn Label"].map({
    "No":0,
    "Yes":1
})
```

```python
# Remove leakage and identifier columns
drop_cols = [
    "CustomerID",
    "Count",
    "Country",
    "State",
    "City",
    "Zip Code",
    "Lat Long",
    "Latitude",
    "Longitude",
    "Churn Value",
    "Churn Score",
    "Churn Reason"
]

X = X.drop(columns=drop_cols)
```

# Train/Test Split

```python
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score, GridSearchCV
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
# Identify categorical and numerical columns
```

```python
numeric_features = X_train.select_dtypes(
    include=np.number
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "category"]
).columns.tolist()

print("Numerical:", numeric_features)
print("Categorical:", categorical_features)
```

```python
# Preprocessing
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])
```

```python
preprocessor = ColumnTransformer([
    ("num", numeric_transformer, numeric_features),
    ("cat", categorical_transformer, categorical_features)
])
```

# Handle imbalance correctly using SMOTE

```python
from imblearn.over_sampling import SMOTE
from sklearn.linear_model import LogisticRegression
```

```python
from imblearn.pipeline import Pipeline as ImbPipeline

logistic_pipeline = ImbPipeline([
    ("preprocessor", preprocessor),
    ("smote", SMOTE(
        random_state=42
    )),
    ("model", LogisticRegression(
        max_iter=1000
    ))
])
```

# Logistic Regression

```python
logistic_model = ImbPipeline([
    ("preprrocessor", preprocessor),
    ("smote", SMOTE(random_state=42)),
    ("model", LogisticRegression(max_iter=1000))
])
logistic_model.fit(X_train, y_train)
```

# Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dt_model = ImbPipeline([
    ("preprocessor", preprocessor),
    ("smote", SMOTE(random_state=42)),
    ("model", DecisionTreeClassifier(
        random_state=42
    ))
])
dt_model.fit(X_train, y_train)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
# Random Forest
rf_model=ImbPipeline([
    ("preprocessor", preprocessor),
    ("smote", SMOTE(random_state=42)),
    ("model", RandomForestClassifier(
        n_estimators=200,
        random_state=42
    ))
])
rf_model.fit(X_train, y_train)
```

# XGBOOST

```python
from xgboost import XGBClassifier
```

```python
xgb_model = ImbPipeline([
    ("preprocessor", preprocessor),
    ("smote", SMOTE(random_state=42)),
    ("model", XGBClassifier(
        n_estimators=200,
        learning_rate=0.05,
        max_depth=4,
        random_state=42,
        eval_metric="logloss",
    ))
])
xgb_model.fit(X_train, y_train)
```

```python
# Evaluate every model
```

```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from sklearn.metrics import roc_curve, roc_auc_score
```

```python
models = {
    "Logistic Regression": logistic_model,
    "Decision Tree": dt_model,
    "Random Forest": rf_model,
    "XGBoost": xgb_model
}
```

```python
results = []

for name, model in models.items():

    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1": f1_score(y_test, y_pred),
        "ROC-AUC": roc_auc_score(y_test, y_prob)
    })

results_df = pd.DataFrame(results)

results_df
```

# ROC Curve

```python
plt.figure(figsize=(8, 6))
for name, model in models.items():
    y_prob = model.predict_proba(X_test)[:,1]
    fqr, tqr, _= roc_curve(
        y_test,
        y_prob
    )
    auc = roc_auc_score(
        y_test,
        y_prob
    )
    plt.plot(
    fqr,
    tqr,
    label=f"{name} AUC={auc:.3f}"
    )
plt.plot(
    [0,1],
    [0,1],
    linestyle="--"
)
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.show()
```

```python
# # Why is accuracy alone not sufficient?
# ----->Accuracy can be misleading when the target is imbalanced. In this dataset, most customers do not churn.
#      A model could achieve relatively high accuracy by predicting many customers as non-churn,
#     while still failing to identify actual churners. Therefore, I also evaluate precision, recall, F1-score and ROC-AUC.
```

```python
def predict_customer_churn(customer_data, model):
    df = pd.DataFrame([customer_data])
    
    # Handle feature aliases if needed
    if "CLTV" in df.columns and "Customer_Lifetime_Value" not in df.columns:
        df["Customer_Lifetime_Value"] = df["CLTV"]
        
    pred = model.predict(df)[0]
    prob = model.predict_proba(df)[0, 1]
    
    return {
        "churn": bool(pred),
        "churn_probability": round(float(prob), 4),
        "risk_level": "High" if prob >= 0.7 else "Medium" if prob >= 0.3 else "Low"
    }

# Test function
result = predict_customer_churn(customer, loaded_model)
print(result)
```
